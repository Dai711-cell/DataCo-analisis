# DataCo Supply Chain Analytics

Este repositorio documenta un proyecto personal de análisis sobre entregas y demanda diaria por producto.

El análisis final y reproducible se desarrolló en scripts Python e informes de `reports/`, no en este cuaderno.

## Hallazgos principales
- Envíos: revisar las promesas de entrega por modalidad antes de añadir modelos más complejos.
- Demanda: la media móvil de 7 días por producto supera a los modelos ML probados con los datos disponibles.
